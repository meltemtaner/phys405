```python
x123 = [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5]
Y = {"I":   [8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68],
     "II":  [9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74],
     "III": [7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73],
     "IV":  [6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89]}
X = {"I": x123, "II": x123, "III": x123, "IV": [8, 8, 8, 8, 8, 8, 8, 19, 8, 8, 8]}
ans = pd.DataFrame([{"dataset": k, "x": xi, "y": yi} for k in Y for xi, yi in zip(X[k], Y[k])])
print(len(ans), "rows")

stats = ans.groupby("dataset").agg(mean_x=("x", "mean"), var_x=("x", "var"),
                                   mean_y=("y", "mean"), var_y=("y", "var"))
stats["r"] = [np.corrcoef(g_.x, g_.y)[0, 1] for _, g_ in ans.groupby("dataset")]
fits = [np.polyfit(g_.x, g_.y, 1) for _, g_ in ans.groupby("dataset")]
stats["slope"] = [f[0] for f in fits]
stats["intercept"] = [f[1] for f in fits]
fmt = {"mean_x": "{:.2f}", "var_x": "{:.2f}", "mean_y": "{:.2f}", "var_y": "{:.3f}",
       "r": "{:.3f}", "slope": "{:.3f}", "intercept": "{:.2f}"}
print(stats.to_string(formatters={k: v.format for k, v in fmt.items()}))

fig, axs = plt.subplots(2, 2, sharex=True, sharey=True, figsize=(8, 6))
xx = np.linspace(3, 20, 2)
for ax, (name, g_) in zip(axs.ravel(), ans.groupby("dataset")):
    ax.plot(g_.x, g_.y, "o", label=f"dataset {name}")
    ax.plot(xx, 3.00 + 0.500 * xx, "-", label="y = 3.00 + 0.500 x")
    ax.legend(loc="upper left")
for ax in axs[1]: ax.set_xlabel("x (arbitrary units)")
for ax in axs[:, 0]: ax.set_ylabel("y (arbitrary units)")
fig.tight_layout()
plt.show()
```

A Ba-137m source is counted for 30 s intervals from t = 0 to 900 s (31 points). **The data below are simulated for this exercise** (seed 405, Poisson counts around a true half-life of 2.552 min = 153.1 s). Estimate the half-life from the table by eye, then fit N = A e^(−t/τ) with `curve_fit(model, t, N)` and no `p0`. What τ does it return?

**Predict.** Write your eye estimate of the half-life (s) and the τ you expect `curve_fit` to return. Write your prediction in the next cell *before* you run anything below. It is a variable, so it is saved with the notebook.

```python
prediction = ...
```

```python
import warnings
from scipy.optimize import curve_fit

T_half = 153.1                      # s, literature Ba-137m (2.552 min)
tau_true = T_half / np.log(2)
tt = np.arange(0, 901, 30.0)        # s
rng = np.random.default_rng(405)
counts = rng.poisson(1000.0 * np.exp(-tt / tau_true)).astype(float)
print(f"{len(tt)} points; first counts {counts[:3].astype(int).tolist()}, last {int(counts[-1])}")

model = lambda t, A, tau: A * np.exp(-t / tau)

# 1) no starting guess: keep the warning visible as text
with warnings.catch_warnings(record=True) as w:
    warnings.simplefilter("always")
    p_bad, c_bad = curve_fit(model, tt, counts)
for x_ in w:
    print("WARNING:", x_.category.__name__, "-", x_.message)
print(f"no p0: A = {p_bad[0]:.0f}, tau = {p_bad[1]:.3f} s, sigma = {np.sqrt(np.diag(c_bad)).tolist()}")
print(f"       model at t = 30 s: {p_bad[0] * np.exp(-30 / p_bad[1]):.1e} counts; data: {int(counts[1])}")

# 2) with a starting guess
p, c = curve_fit(model, tt, counts, p0=(counts[0], 200.0))
sA, stau = np.sqrt(np.diag(c))
print(f"p0 = (counts[0], 200 s): A = {p[0]:.0f} +/- {sA:.0f}, tau = {p[1]:.1f} +/- {stau:.1f} s")
print(f"T_half = tau ln2 = {p[1]*np.log(2):.1f} +/- {stau*np.log(2):.1f} s = {p[1]*np.log(2)/60:.2f} min (literature 2.552 min)")

fig, ax = plt.subplots(figsize=(7, 4.5))
ax.errorbar(tt, counts, yerr=np.sqrt(counts), fmt="o", capsize=2, label="counts (error bars: sqrt(N))")
tf = np.linspace(0, 900, 300)
ax.plot(tf, model(tf, *p), "-", label=f"fit with p0: tau = {p[1]:.1f} s")
ax.plot(tf, model(tf, *p_bad), "--", label=f"fit without p0: tau = {p_bad[1]:.3f} s")
ax.set_yscale("log"); ax.set_ylim(1, 3000)
ax.set_xlabel("time t (s)"); ax.set_ylabel("counts per 30 s interval")
ax.legend()
plt.show()
```
